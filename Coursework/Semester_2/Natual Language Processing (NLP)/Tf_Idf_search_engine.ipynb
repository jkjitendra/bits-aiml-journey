```python
import math

# 1. The Real-World Corpus (Same as before)
corpus = [
    "Apple announced a new iPhone with a better camera and longer battery life.",
    "The stock market is volatile; investors are advised to diversify their portfolios.",
    "Doctors recommend a balanced diet and regular exercise for heart health.",
    "Python is a versatile programming language used for data science and web development.",
    "The Federal Reserve raised interest rates to combat rising inflation in the market."
]
```

```python
# 2. Preprocessing (Cleaning the text manually)
# We need to remove punctuation and common words ("stop words") to make it work well.
STOP_WORDS = set(["the", "a", "an", "and", "is", "to", "in", "for", "of", "with"])

def clean_and_tokenize(sentence):
    # Remove punctuation manually
    for char in ".,;":
        sentence = sentence.replace(char, "")

    # Split and lower case
    tokens = sentence.lower().split()

    # Remove stop words
    filtered_tokens = [word for word in tokens if word not in STOP_WORDS]
    return filtered_tokens

# Pre-process all documents
cleaned_corpus = [clean_and_tokenize(doc) for doc in corpus]
```

```python
# 3. TF Calculation
# Formula: (Count of term t in doc) / (Total words in doc)
def compute_tf(term, tokenized_doc):
    count = tokenized_doc.count(term)
    if len(tokenized_doc) == 0: return 0
    return count / len(tokenized_doc)

# 4. IDF Calculation
# Formula: log(Total Docs / Docs with term t)
def compute_idf(term, all_tokenized_docs):
    # Count how many docs contain the term
    count_containing = sum(1 for doc in all_tokenized_docs if term in doc)

    # If word is not in any document (e.g. user searches for "alien"), return 0
    if count_containing == 0: return 0

    # Standard IDF formula (+1 is sometimes added to denominator to avoid division by zero,
    # but for this simple demo, we assume the word exists if we are calculating it)
    return math.log(len(all_tokenized_docs) / count_containing)
```

```python
# 5. The Search Engine Logic
def search(query):
    print(f"\n--- Searching for: '{query}' ---")
    query_tokens = clean_and_tokenize(query)

    # We will store (Score, Document) tuples
    results = []

    for i, doc_tokens in enumerate(cleaned_corpus):
        score = 0

        # For every word in the user's query, calculate TF-IDF
        for word in query_tokens:
            tf = compute_tf(word, doc_tokens)
            idf = compute_idf(word, cleaned_corpus)
            weight = tf * idf
            score += weight

            # Optional: Print the math for students to see!
            if weight > 0:
                print(f"   [Match: '{word}'] in Doc {i} -> TF: {tf:.2f} * IDF: {idf:.2f} = {weight:.2f}")

        results.append((score, corpus[i]))

    # Sort results by score (Highest first)
    results.sort(key=lambda x: x[0], reverse=True)

    # Display Top Results
    found = False
    for score, doc in results:
        if score > 0:
            print(f"Score: {score:.4f} | {doc}")
            found = True

    if not found:
        print("No relevant results found.")
```
