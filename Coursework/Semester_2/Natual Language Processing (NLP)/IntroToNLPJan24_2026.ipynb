```python
from textblob import TextBlob

# Step 1: The Input (Real, messy human language)
# Notice the complexity: It starts positive, uses a connector ('but'),
# and ends with a strong negative idiom ('nightmare').
review = "The flight attendants were super friendly, but the food was garbage and the delay was a nightmare."

# Step 2: Create the Blob (The Processing)
blob = TextBlob(review)

# Step 3: Extract Sentiment
polarity = blob.sentiment.polarity
# Polarity ranges from -1.0 (Very Negative) to +1.0 (Very Positive)

print(f"Review: {review}")
print(f"Sentiment Score: {polarity}")
```

```python
# The "Hello World" of NLP
from textblob import TextBlob

# Scenario: A customer leaves a review for a hotel.
review_1 = "The hotel was clean and the staff was friendly."
review_2 = "The hotel was clean, but the staff was rude and the food was garbage."

# Let's analyze Review 1
blob1 = TextBlob(review_1)
print(f"Review 1 Sentiment: {blob1.sentiment.polarity}")
# Output: +0.6 (Positive)

# Let's analyze Review 2
blob2 = TextBlob(review_2)
print(f"Review 2 Sentiment: {blob2.sentiment.polarity}")
# Output: -0.2 (Negative)
```

```python
import spacy
# Load the "Brain" (The English language model)
nlp = spacy.load("en_core_web_sm")

# The Raw Data
text = "I hated the flight. The food was cold, but the staff was friendly."
```

```python
# This one line runs Phonology, Morphology, Syntax, and Semantics!
doc = nlp(text)

# Let's look at Morphology (Tokenization & POS Tagging)
print(f"{'Text':<10} | {'POS':<10} | {'Lemma'}")
print("-" * 30)
for token in doc:
    print(f"{token.text:<10} | {token.pos_:<10} | {token.lemma_}")
```

```python
text_2 = "Google was founded by Larry Page in California in 1998."
doc_2 = nlp(text_2)

for ent in doc_2.ents:
    print(f"{ent.text} -> {ent.label_}")
```

```python
# First, you might need to install these libraries:
# pip install scikit-learn nltk

from sklearn.metrics import precision_score, recall_score, f1_score
from nltk.translate.bleu_score import sentence_bleu

print("--- PART 1: CLASSIFICATION METRICS (Spam Detection) ---")

# SCENARIO: 1 = Spam, 0 = Not Spam
# y_true: The Ground Truth (What the emails actually were)
y_true = [1, 0, 1, 1, 0, 1]

# y_pred: The Model's Guess (What the AI thought they were)
y_pred = [1, 0, 0, 1, 1, 1]

# 1. Precision: When the model predicted 'Spam' (1), how often was it right?
# The model predicted '1' four times (indices 0, 3, 4, 5).
# It was correct 3 times (0, 3, 5) and wrong once (4).
precision = precision_score(y_true, y_pred)

# 2. Recall: Out of all actual 'Spam' emails, how many did it find?
# There were 4 actual spam emails (indices 0, 2, 3, 5).
# The model found 3 of them (0, 3, 5) and missed one (2).
recall = recall_score(y_true, y_pred)

# 3. F1 Score: The harmonic mean (balance) of the two.
f1 = f1_score(y_true, y_pred)

print(f"Ground Truth: {y_true}")
print(f"Predictions:  {y_pred}")
print(f"Precision:    {precision:.2f} (Model is 75% trustworthy when it flags spam)")
print(f"Recall:       {recall:.2f}    (Model found 75% of the real spam)")
print(f"F1 Score:     {f1:.2f}        (The overall balance)")


print("\n--- PART 2: GENERATION METRICS (Machine Translation) ---")

# SCENARIO: Translating a French sentence to English.

# The Reference (Human Translation - Ground Truth)
# NLTK expects a list of references (because there can be multiple valid translations)
reference = [['the', 'cat', 'is', 'on', 'the', 'mat']]

# The Candidate (Machine Translation)
# The machine missed the word "the"
candidate = ['the', 'cat', 'is', 'on', 'mat']

# 4. BLEU Score
# It checks n-gram overlap.
# 1-gram match: 'the', 'cat', 'is', 'on', 'mat' (5/5 matches found in reference)
# But it penalizes for brevity (the candidate is too short).
score = sentence_bleu(reference, candidate)

print(f"Reference (Human): {' '.join(reference[0])}")
print(f"Candidate (AI):    {' '.join(candidate)}")
print(f"BLEU Score:        {score:.4f} (Perfect match would be 1.0)")
```
